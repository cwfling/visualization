```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100,50)

# Generating a scatter plot, bar plot, line plot, and histogram using the same data
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y)

fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x,y)

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)

fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)

# Labeling the axes and title
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout()

# Setting font styles for the title and axis labels using font dictionaries
font1 = {'family':'sans-serif','color':'blue','size':20}
font2 = {'family':'monospace','color':'green','size':14}

# Using the font dictionaries to set the font properties of the title and axis labels
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict = font1)
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
fig.tight_layout()

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,
         y,
         marker='*', # NEW ADDITIONS
         color = "indigo") # NEW ADDITIONS
fig.show()

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,
y,
marker='*',
color = 'indigo',
linestyle = '--', # NEW ADDITIONS
linewidth = 2) # NEW ADDITIONS
fig.show()

fig, ax = plt.subplots(figsize=(5, 3))

# color being the new addition
ax.plot(x,y,marker='*', color = '#7425b9', linestyle = '--', linewidth = 2)
fig.show()

# Changing the marker edge color and marker face color to make the markers more visible
ax.plot(x,y,marker='*',
markersize = 12, # NEW ADDITIONS
color = '#7425b9',
linestyle = '--',
linewidth = 2,
markeredgecolor = '#fa9359', # NEW ADDITIONS
markerfacecolor = '#000000' ) # NEW ADDITIONS

ax.grid(axis = 'y')
ax.grid(axis = 'y', color = "blue", linewidth = 2, linestyle = '-.')
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0, 100,50)
y2 = np.random.randint(0, 100,50)
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='lower right',
frameon = True, #add frame around the legend
fontsize = 12, #change font size
ncol = 2, #specify number of columns
shadow = True )
fig.show()

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='upper left',
bbox_to_anchor =(1, 1))
fig.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.text(10, 95, "This value is important!",
ha='center', ##alignment of text
color = 'red', ##modify font colour
size = 20) ##modify font size
fig.show()
```

```python
fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])
ax.text(1, 5, ". Data:(1, 5)", transform=ax.transData )
ax.text(0.5, 0.1, ". Axes:(0.5, 0.1)", transform=ax.transAxes )
ax.text(0.2, 0.2, ". Figure:(0.2, 0.2)", transform=fig.transFigure )

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!', xy=(10, 95), xytext=(20, 94),
arrowprops=dict(facecolor='black'))
fig.show()
ax.annotate('This is important!',
xy=(10, 95), xytext=(20, 94),
arrowprops = dict(arrowstyle = "->",
color = "red"))
fig.show()

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.yaxis.set_major_locator(plt.NullLocator())
ax.xaxis.set_major_formatter(plt.NullFormatter())

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MaxNLocator(3))

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MultipleLocator(5))

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
plt.xticks(rotation=45, ha='right')
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])

fig, ax = plt.subplots(figsize=(5, 3))
fig,(ax1, ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7, 3))
fig, (ax1, ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7, 3))
ax1.scatter(x1,y1)
ax2.bar(x2,y2)
fig.show()
```

```python
font1 = {'family':'serif','color':'red'}
fig, (ax1, ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7, 3))
ax2.bar(x2,y2)
ax2.legend(loc='upper right')
ax2.set_xlabel('One Piece Characters', fontsize = 15, fontdict = font1)
ax2.set_ylabel('Power Level', fontsize = 15, fontdict = font1)
fig.show()

fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'], ['ax2', 'ax3']], figsize=(7, 4))

fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'], ['ax2', 'ax3']], figsize=(7, 4))
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
plt.show()
```

```python
y2_sd = np.std(y2)
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2, color = "red")
ax.errorbar(x, #our x values
y2, #our y values
yerr = y2_sd,
fmt = "none")

fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y2,
color = "red")
ax.errorbar(x,
y2,
yerr = y2_sd,
fmt = "none",
ecolor= "indigo",
elinewidth= 4,
capsize = 6,
capthick= 4 )

ax.errorbar(x,
y2,
yerr = y2_sd,
fmt = "none",
ecolor = "indigo",
elinewidth = 4,
capsize = 6,
capthick = 4,
errorevery= 2 )
```

```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests

tips = sns.load_dataset("tips")
print(tips)

sns.lineplot(data=tips, # choose our dataset
x='total_bill', # define our x variable
y='tip') # define our y variable

sns.set_style('whitegrid')

tipgraph = sns.lineplot(data=tips,
x='total_bill',
y='tip')
tipgraph.set(title='Tips vs. Total Bill',
xlabel='Total Bill ($)',
ylabel='Tip Amount ($)')

fig = plt.subplots(figsize=(10, 3))
tipgraph = sns.lineplot(data=tips,
x='total_bill',
y='tip',
color = 'hotpink',
linestyle = '--',
linewidth = 3,
marker = 'o',
markerfacecolor = 'indigo')

```

```python
tipgraph = sns.scatterplot(data=tips, x='total_bill',
y='tip', style = 'time', hue =
'day', palette = ['purple',
'hotpink', 'deepskyblue',
'yellowgreen'])
tipgraph.set(title='Tips vs. Total Bill',
xlabel='Total Bill ($)',
ylabel='Tip Amount ($)')

sns.pairplot(
data = tips,
hue = 'day')
```

```python
daysplot = sns.relplot(
data=tips, # grabs the dataset
x="total_bill", # sets the x variable
y="tip", # sets the y variable
hue="sex", # adjusts the colour of the points based on the sex category from the dataset
col="day", # adjusts the plots based on the day category from the dataset
kind="scatter", # sets the type of plot to scatterplot
col_wrap=2) # wraps the column plots into 2 columns
```

Left plot is greyscale, looks boring and standard. Very high level overview of the data with sampling information provided

Right side is more engaging. Colours have meaning and show the variance on a finer level using the categories of subject. Shows finer details via the sample points between and standard deviation. Harder to tell how many samples were used and what averages are but gives more context with other variables

```python
import plotly.graph_objects as go # 'go' is 'graph objects’
x1 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])

graph = go.Figure()

graph.add_trace(go.Bar(x=x1, y=y1))

graph.update_layout(
title="Pirate Scores",
xaxis_title="Pirates",
yaxis_title="Score")

graph.show()

graph.write_html("../../folders/pirategraph.html")

graph = go.Figure()
graph.add_trace(go.Scatter(x=x1, y=y1, mode='markers', # we want points for a scatter plot
marker=dict(
size=15, # point size
color='hotpink', # point colour
opacity=1, # point transparency/alpha
line=dict(width=5, color='purple') # point outline
)))
graph.update_layout(
title='Interactive Pirate Plot',
xaxis_title='Pirates',
yaxis_title='Scores',
width=500, height=500)
```
