The data was obtained from the TTC Streetcar Delay data on the City of Toronto Data page (https://open.toronto.ca/dataset/ttc-streetcar-delay-data/). It was selected due to relevance of the data and a relatively high data score and grade of 83% and gold respectively. 

```python
#Import .csv file into pandas dataframe and display the data
import pandas as pd
df = pd.read_csv('C:\\Users\\chris\\OneDrive - University of Toronto\\Documents\\PhD Work\\Courses\\Data Science Course\\visualization\\02_activities\\assignments\\assignment 3\\TTC Streetcar Delay Data since 2025.csv')
print(df)
```

```python
#Show the column names in the data
print(df.columns)

#Show the number of unique routes in the data
print(df['Line'].nunique())

#Group by the route Line, calculate average delay time, sort from highest to lowest, and display the result
average_delay = df.groupby('Line')['Min Delay'].mean().sort_values(ascending=False)
print(average_delay)

```

```python
#Create a histogram of the delay times
import matplotlib.pyplot as plt
#Create custom bins to make histogram more readable
bins = list(range(0, 190, 15))
#Axis formatting
plt.xlabel('Delay Time (Minutes)')
plt.ylabel('Frequency')
plt.title('Histogram of Streetcar Delay Times')
plt.xticks(bins)
counts, edges, bars = plt.hist(average_delay, bins=bins, edgecolor='black')
#Hide the 0 value labels above the histogram bars
plt.bar_label(bars, labels=[int(i) if i > 0 else '' for i in bars.datavalues])
#Colour the bins above 30 minutes in red
for bar, edge in zip(bars, edges):
    if edge >= 30:
        bar.set_color('red')
        bar.set_edgecolor('black')
#Create dotted vertical line at 30 minute mark
plt.axvline(x=30, color='red', linestyle='dotted')
plt.legend(['30 Minute Delay Threshold'])
#Save the histogram as a .png file
plt.savefig('streetcar_delay_histogram.png')
```
